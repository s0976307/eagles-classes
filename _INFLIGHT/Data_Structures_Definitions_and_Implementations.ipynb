## Doubly Linked List

A doubly linked list contains nodes.

Each node has:

- `data` - the value stored in the node
- `next` - points to the next node
- `prev` - points to the previous node

The list keeps track of:

- `head` - the first node
- `tail` - the last node

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def insertAtTail(self, node):
        if self.head is None:
            self.head = node
            self.tail = node
        else:
            node.prev = self.tail
            self.tail.next = node
            self.tail = node

    def deleteAtHead(self, node):
        if self.head is None:
            return

        self.head = self.head.next

        if self.head is None:
            self.tail = None
        else:
            self.head.prev = None

    def printHeadToTailData(self):
        current = self.head

        while current is not None:
            print(current.data)
            current = current.next
```

```python
### Test the Doubly Linked List
dll = DoublyLinkedList()

dll.insertAtTail(Node(1))
dll.insertAtTail(Node(2))
dll.insertAtTail(Node(3))

dll.printHeadToTailData()
```

## Array

An array stores values using indexes.

For example:

Index:  0   1   2   3
Value: 10  20  30  40

The `set()` method changes a value at an index.

The `get()` method returns the value at an index.

```python
class Array:
    def __init__(self, N):
        self.data = [None] * N

    def set(self, i, data):
        self.data[i] = data

    def get(self, i):
        return self.data[i]
```

```python
### Test the Array
arr = Array(5)

arr.set(0, 10)
arr.set(1, 20)
arr.set(2, 30)

print(arr.get(0))
print(arr.get(1))
print(arr.get(2))
```

### Time Complexity

The `set()` method has a time complexity of **O(1)**.

The `get()` method has a time complexity of **O(1)**.

This is because the program can access an array index directly.

Python's `list.append()` function has an **O(1) amortized** time complexity.

This means that most append operations take O(1) time. Occasionally Python must resize the list, which can take O(n), but the average cost remains O(1).

Therefore:

- `set()` = O(1)
- `get()` = O(1)
- `list.append()` = O(1) amortized

The main difference is that `set()` changes an existing index, `get()` retrieves an existing index, and `append()` adds a new value to the end of a list.

## Graph

A graph contains vertices and connections called edges.

For this assignment, the graph is **non-directed**.

We will use an **adjacency matrix**.

The matrix uses:

- `True` = a connection exists
- `False` = no connection exists

The `labels` list contains the names of the vertices.

```python
class Graph:
    def __init__(self, labels):
        self.labels = labels

        n = len(labels)
        self.matrix = [[False] * n for _ in range(n)]

    def hasConnection(self, vertexAlabel, vertexBlabel):
        a = self.labels.index(vertexAlabel)
        b = self.labels.index(vertexBlabel)

        return self.matrix[a][b]
```

```python
### Test the Graph
g = Graph(["A", "B", "C"])

# A connects to B
g.matrix[0][1] = True
g.matrix[1][0] = True

# B connects to C
g.matrix[1][2] = True
g.matrix[2][1] = True

print(g.hasConnection("A", "B"))
print(g.hasConnection("B", "C"))
print(g.hasConnection("A", "C"))
```
